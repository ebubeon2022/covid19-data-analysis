## 1. Import the libraries and load the CSV file

**Question:** Import the COVID-19 CSV file into a DataFrame.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("COVID-19 Cases World.csv")

print("COVID-19 dataset imported successfully!")
print(f"Number of rows: {df.shape[0]:,}")
print(f"Number of columns: {df.shape[1]}")
display(df.head())
```

## 2. Keep only the data required for the analysis

**Question:** Isolate the required data, such as dates, cases, deaths, and countries.

The notebook also keeps the continent, month and year columns because later questions use them.

```python
required_columns = [
    "dateRep",
    "day",
    "month",
    "year",
    "cases",
    "deaths",
    "countriesAndTerritories",
    "continentExp"
]

df = df[required_columns].copy()

# Convert the date column to a date format.
df["dateRep"] = pd.to_datetime(df["dateRep"], dayfirst=True, errors="coerce")

# Make sure the case and death columns are numeric.
df["cases"] = pd.to_numeric(df["cases"], errors="coerce").fillna(0)
df["deaths"] = pd.to_numeric(df["deaths"], errors="coerce").fillna(0)

print("Required columns selected successfully.")
display(df.head())
```

## 3. Summarise worldwide COVID-19 cases and deaths

**Question:** Give a summary of the total number of cases and deaths worldwide.

```python
worldwide_cases = int(df["cases"].sum())
worldwide_deaths = int(df["deaths"].sum())

print(f"Total reported COVID-19 cases worldwide: {worldwide_cases:,}")
print(f"Total reported COVID-19 deaths worldwide: {worldwide_deaths:,}")
```

## 4. Visualise the total number of cases by country

**Question:** Use a bar chart to visualise the total number of cases by countries.

For readability, the chart displays the 10 countries with the highest total reported cases in this dataset.

```python
cases_by_country = (
    df.groupby("countriesAndTerritories")["cases"]
    .sum()
    .sort_values(ascending=False)
)

top_10_countries = cases_by_country.head(10)

plt.figure(figsize=(12, 6))
top_10_countries.sort_values().plot(kind="barh")
plt.title("Top 10 Countries by Total Reported COVID-19 Cases")
plt.xlabel("Total reported cases")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

## 5. Sum the number of cases worldwide

**Question:** Sum the number of cases worldwide.

```python
total_cases_worldwide = int(df["cases"].sum())
print(f"The total number of reported COVID-19 cases worldwide is {total_cases_worldwide:,}.")
```

## 6. Sum the number of cases per continent

**Question:** Provide the sum of cases per continent.

```python
cases_by_continent = (
    df.groupby("continentExp")["cases"]
    .sum()
    .sort_values(ascending=False)
)

print("Total reported COVID-19 cases by continent:")
display(cases_by_continent.rename("total_cases").to_frame())
```

## 7. Sum the deaths by continent

**Question:** Provide the sum of the death rates by continent.

**Data note:** The supplied CSV contains a `deaths` column (reported death counts), not a separate death-rate column. This solution therefore sums the reported deaths by continent rather than calculating a death rate.

```python
deaths_by_continent = (
    df.groupby("continentExp")["deaths"]
    .sum()
    .sort_values(ascending=False)
)

print("Total reported COVID-19 deaths by continent:")
display(deaths_by_continent.rename("total_deaths").to_frame())
```

## 8. Sum the number of cases in the United Kingdom

**Question:** Sum the number of cases in the United Kingdom.

```python
uk_rows = df[df["countriesAndTerritories"] == "United_Kingdom"]
uk_total_cases = int(uk_rows["cases"].sum())

print(f"Total reported COVID-19 cases in the United Kingdom: {uk_total_cases:,}")
```

## 9. Sum the number of cases in Canada

**Question:** Sum the number of cases in Canada.

```python
canada_rows = df[df["countriesAndTerritories"] == "Canada"]
canada_total_cases = int(canada_rows["cases"].sum())

print(f"Total reported COVID-19 cases in Canada: {canada_total_cases:,}")
```

## 10. Compare cases in June 2020 in Canada and the United Kingdom

**Question:** Compare the number of cases in June in Canada and the United Kingdom in 2020.

```python
june_2020 = df[(df["month"] == 6) & (df["year"] == 2020)]

june_comparison = (
    june_2020[
        june_2020["countriesAndTerritories"].isin(["Canada", "United_Kingdom"])
    ]
    .groupby("countriesAndTerritories")["cases"]
    .sum()
)

uk_june_cases = int(june_comparison.get("United_Kingdom", 0))
canada_june_cases = int(june_comparison.get("Canada", 0))

print("COVID-19 cases reported during June 2020:")
print(f"United Kingdom: {uk_june_cases:,}")
print(f"Canada: {canada_june_cases:,}")

if uk_june_cases > canada_june_cases:
    print(f"The United Kingdom reported {uk_june_cases - canada_june_cases:,} more cases than Canada in June 2020.")
elif canada_june_cases > uk_june_cases:
    print(f"Canada reported {canada_june_cases - uk_june_cases:,} more cases than the United Kingdom in June 2020.")
else:
    print("Both countries reported the same number of cases in June 2020.")
```

## 11. Display a user-friendly summary

**Question:** Output the result to the command line with a user-friendly message.

```python
print("=" * 55)
print("COVID-19 DATA ANALYSIS SUMMARY")
print("=" * 55)
print(f"Worldwide reported cases: {worldwide_cases:,}")
print(f"Worldwide reported deaths: {worldwide_deaths:,}")
print(f"United Kingdom reported cases: {uk_total_cases:,}")
print(f"Canada reported cases: {canada_total_cases:,}")
print("\nJune 2020 comparison:")
print(f"  United Kingdom: {uk_june_cases:,} cases")
print(f"  Canada: {canada_june_cases:,} cases")
print("=" * 55)
print("Analysis completed successfully.")
```
